```python
import pandas as pd
```

```python
data=pd.read_csv(r'D:\New folder\uncleaned_sales_data (1).csv')
```

```python
data
```

```python
data.info()
```

```python

text_cols = ['customer_name', 'email', 'product', 'category', 'city', 'payment_method']
for col in text_cols:
    data[col] = data[col].astype(str).str.strip()
```

```python

data['category'] = data['category'].str.title()
data['city'] = data['city'].str.title()
```

```python
data['payment_method'] = data['payment_method'].str.lower().map({
    'credit card': 'Credit Card',
    'debit card': 'Debit Card',
    'paypal': 'PayPal',
    'cash': 'Cash'
})
```

```python
data['customer_name'] = data['customer_name'].replace({'N/A': pd.NA, 'nan': pd.NA})
data['email'] = data['email'].replace('nan', pd.NA)
data['product'] = data['product'].replace('nan', pd.NA)
data['city'] = data['city'].replace('Nan', pd.NA)
data['order_date'] = data['order_date'].replace('not available', pd.NA)
```

```python
before_dupes = len(data)
data = data.drop_duplicates(subset=['customer_name', 'email', 'product', 'order_date'], keep='first')
removed_dupes = before_dupes - len(data)
```

```python
data['order_date'] = pd.to_datetime(data['order_date'], format='mixed', errors='coerce')
```

```python
data['quantity'] = pd.to_numeric(data['quantity'], errors='coerce')
data['price'] = pd.to_numeric(data['price'], errors='coerce')
```

```python
data.loc[data['price'] < 0, 'price'] = data['price'].abs()
```

```python
data.loc[data['quantity'] > 50, 'quantity'] = pd.NA
```

```python
data['price'] = data['price'].fillna(data['price'].median())
data['quantity'] = data['quantity'].fillna(data['quantity'].median())
```

```python
data['category'] = data['category'].fillna('Unknown')
data['city'] = data['city'].fillna('Unknown')
data['email'] = data['email'].fillna('unknown@email.com')
data['customer_name'] = data['customer_name'].fillna('Unknown')
data['product'] = data['product'].fillna('Unknown')
```

```python
data['total_amount'] = data['quantity'] * data['price']
```

```python
assert data['price'].min() >= 0, "Negative prices still present!"
assert data.duplicated(subset=['customer_name','email','product','order_date']).sum() == 0, "Duplicates still present!"
```

```python
data.to_csv('cleaned_sales_data.csv', index=False)
```
